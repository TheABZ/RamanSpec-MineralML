# 3. PCA analysis

Distinguish sample scores from spectral loadings.

All spectra are synthetic; class IDs are not mineral assignments. Run cells in order using the installed project environment.


Fit using observations only, after preprocessing.

```python
import numpy as np
import matplotlib.pyplot as plt
from ramanspec.simulation import simulate
from ramanspec.preprocessing import preprocess
from ramanspec.dimensionality import run_pca
ds = simulate()
x, diagnostics = preprocess(ds.observed)
scores, pca = run_pca(x, 5)
print(scores.shape, pca.components_.shape)
print(pca.explained_variance_ratio_)
```

Truth colors are diagnostic, not PCA inputs.

```python
fig, ax = plt.subplots()
ax.scatter(scores[:,0], scores[:,1], c=ds.labels, s=8)
ax.set(xlabel="PC1 score", ylabel="PC2 score", title="Synthetic class colors")
plt.show()
```

Exercise: compare normalized and standardized PCA. Which spectral channels dominate?

```python
fig, ax = plt.subplots()
for j in range(3):
    ax.plot(ds.axis, pca.components_[j], label=f"PC{j+1}")
ax.set(xlabel="Raman shift (cm⁻¹)", ylabel="Loading")
ax.legend()
plt.show()
```
