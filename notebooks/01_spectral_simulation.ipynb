# 1. Spectral simulation

Explore peak parameters, array shapes and ground-truth decomposition.

All spectra are synthetic; class IDs are not mineral assignments. Run cells in order using the installed project environment.


Generate the default dataset. Rows are pixels and columns are spectral channels.

```python
import numpy as np
import matplotlib.pyplot as plt
from ramanspec.simulation import simulate, SimulationConfig, gaussian
ds = simulate()
print(ds.observed.shape, ds.axis[[0,-1]])
assert np.allclose(ds.observed, ds.clean + ds.baseline + ds.noise + ds.spikes)
```

Inspect one invented class signature and the contamination terms.

```python
fig, ax = plt.subplots(figsize=(9,4))
for name in ["clean", "baseline", "observed"]:
    ax.plot(ds.axis, getattr(ds,name)[0], label=name)
ax.set(xlabel="Raman shift (cm⁻¹)", ylabel="Intensity (a.u.)")
ax.legend()
plt.show()
```

Exercise: change the seed and widths. What remains reproducible?

```python
again = simulate(SimulationConfig(seed=42))
assert np.array_equal(ds.observed, again.observed)
print("Same seed reproduces all observations")
```
