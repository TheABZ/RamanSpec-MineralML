# 4. Synthetic spectral clustering

The filename anticipates future mineral applications; this notebook uses invented classes only.

All spectra are synthetic; class IDs are not mineral assignments. Run cells in order using the installed project environment.


Fit k-Means and GMM without labels or coordinates.

```python
import matplotlib.pyplot as plt
from ramanspec.simulation import simulate
from ramanspec.preprocessing import preprocess
from ramanspec.dimensionality import run_pca
from ramanspec.clustering import cluster, model_selection
from ramanspec.validation import evaluate_clusters, align_labels
ds = simulate()
scores, pca = run_pca(preprocess(ds.observed)[0])
result = cluster(scores)
for name in ["kmeans", "gmm"]:
    print(name, evaluate_clusters(ds.labels, result[name], scores))
```

Align labels for colors only. Exercise: permute IDs and confirm ARI is unchanged.

```python
fig, axes = plt.subplots(1,3,figsize=(10,3))
for ax, labels, title in zip(axes, [ds.labels, align_labels(ds.labels,result["kmeans"]), align_labels(ds.labels,result["gmm"])], ["Synthetic truth", "k-Means", "GMM"]):
    ax.imshow(labels.reshape(30,30), vmin=0, vmax=2)
    ax.set(title=title, xlabel="Column (pixel)", ylabel="Row (pixel)")
plt.tight_layout()
plt.show()
```
