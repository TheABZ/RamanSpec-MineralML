# 2. Baseline correction

Inspect robust despiking, sparse AsLS and baseline error.

All spectra are synthetic; class IDs are not mineral assignments. Run cells in order using the installed project environment.


Despike one observation; fit the baseline without providing clean truth.

```python
import matplotlib.pyplot as plt
from ramanspec.simulation import simulate
from ramanspec.despiking import despike
from ramanspec.baseline import asls
from ramanspec.validation import rmse
ds = simulate()
y, flags = despike(ds.observed[0])
z, info = asls(y, return_info=True)
print(info, "Baseline RMSE:", rmse(ds.baseline[0], z))
```

Compare truth only after fitting. Exercise: vary p and λ and explain the effect.

```python
fig, ax = plt.subplots(figsize=(9,4))
for values, label in [(ds.observed[0], "Observed"), (ds.baseline[0], "True baseline"), (z, "AsLS"), (y-z, "Corrected")]:
    ax.plot(ds.axis, values, label=label)
ax.set(xlabel="Raman shift (cm⁻¹)", ylabel="Intensity (a.u.)")
ax.legend()
plt.show()
```

A single example does not establish population accuracy.

```python
for lam in [1e4,1e5,1e6]:
    print(lam, rmse(ds.baseline[0], asls(y, lam=lam)))
```
